### **Exercise 3: Slice the data from '2013-01-01' to '2013-12-31'**
**Question:** Slice the DataFrame to show data for the year 2013.

```python
# Your code here:
import pandas as pd

# Load dataset again
file_path = "https://raw.githubusercontent.com/Hamed-Ahmadinia/DASP-2025/main/Bike%20Sales.csv"
df = pd.read_csv(file_path)

# Check column names to identify any extra spaces
print("Column Names:", df.columns)

# Trim spaces from column names if needed
df.columns = df.columns.str.strip()

# Check if "Date" exists
if "Date" in df.columns:
    # Convert "Date" column to datetime format
    df["Date"] = pd.to_datetime(df["Date"])

    # Set "Date" as the index
    df.set_index("Date", inplace=True)

    # Sort the index
    df = df.sort_index()

    # Slice data for the year 2013
    df_2013 = df.loc["2013-01-01":"2013-12-31"]

    # Display results
    print(df_2013.head())
    print(f"Total records in 2013: {df_2013.shape[0]}")
else:
    print("Error: 'Date' column not found in DataFrame.")

```

### **Exercise 4: Calculate cumulative revenue**
**Question:** Add a new column 'Cumulative_Revenue' that shows the cumulative sum of the revenue.

```python
# Your code here:
# Ensure that 'Revenue' column exists
if "Revenue" in df.columns:
    # Calculate cumulative revenue
    df["Cumulative_Revenue"] = df["Revenue"].cumsum()

    # Display the first few rows to verify
    print(df[["Revenue", "Cumulative_Revenue"]].head(10))  # Show first 10 rows
else:
    print("Error: 'Revenue' column not found in DataFrame.")
```

### **Exercise 5: Downsample to show monthly total revenue (Hint: Use resampling)**
**Question:** Resample the data to calculate total monthly revenue.

```python
# Your code here:
# Resample the data by month (using 'ME' for month end) and calculate total revenue for each month
monthly_revenue = df.resample('ME').agg({'Revenue': 'sum'})

# Display the first few rows to verify
print(monthly_revenue.head())  # Show first 5 months of total revenue

```
