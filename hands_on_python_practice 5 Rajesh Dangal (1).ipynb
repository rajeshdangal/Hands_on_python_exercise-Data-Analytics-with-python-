### **Question 2: Compute the Covariance Between Consecutive Days**
- Create a new column: **Previous Day Views"**, which contains the previous day's pageviews.
- Compute the covariance between **"value"** (current day's views) and **"Previous Day Views"**.


```python
# Your code here:
# Create a new column with previous day's pageviews
df["Previous Day Views"] = df["value"].shift(1)

df_cleaned = df.dropna()

covariance = df_cleaned[["value", "Previous Day Views"]].cov().iloc[0, 1]

print("Covariance between current day's views and previous day's views:", covariance)
```

### **Question 3: Compute Pearson, Spearman, and Kendall Correlation Between Pageviews and Previous Day Views**
- Compute and compare three different correlation measures.?

```python
# Your code here:
from scipy.stats import pearsonr, spearmanr, kendalltau

df["Previous Day Views"] = df["value"].shift(1)
df_cleaned = df.dropna()

current = df_cleaned["value"]
previous = df_cleaned["Previous Day Views"]

pearson_corr, _ = pearsonr(current, previous)
spearman_corr, _ = spearmanr(current, previous)
kendall_corr, _ = kendalltau(current, previous)

# Display results
print("Correlation between 'value' and 'Previous Day Views':")
print(f"Pearson correlation:  {pearson_corr:.4f}")
print(f"Spearman correlation: {spearman_corr:.4f}")
print(f"Kendall correlation:  {kendall_corr:.4f}")
```

### **Question 6: Test Whether Pageviews Have Increased Over Time**
- Compare mean pageviews in the first 30 days vs. the last 30 days.
- Perform a one-sample T-test

```python
# Your code here:
from scipy.stats import ttest_ind

first_30_days = df.iloc[:30]["value"]
last_30_days = df.iloc[-30:]["value"]

mean_first_30 = first_30_days.mean()
mean_last_30 = last_30_days.mean()

t_stat, p_value = ttest_ind(first_30_days, last_30_days)

print("Comparison of Pageviews (First 30 Days vs Last 30 Days):")
print(f"Mean of first 30 days: {mean_first_30:.2f}")
print(f"Mean of last 30 days: {mean_last_30:.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("The difference in pageviews is statistically significant.")
else:
    print("The difference in pageviews is not statistically significant.")
```
