### **Exercise 3: Lists in Python**
**Question:** Create a list of the unique values in the `education` column and print its length.

```python
import pandas as pd

data = {'education': ['Bachelors', 'Masters', 'Bachelors', 'PhD', 'High School']}
df = pd.DataFrame(data)

unique_education = df['education'].unique().tolist()

print(len(unique_education))

education_list = ['Bachelors', 'Masters', 'Bachelors', 'PhD', 'High School']
unique_education = list(set(education_list))

print(len(unique_education))
```

### **Exercise 4: Sets in Python**
**Question:** Check if the value "Self-employed" exists in the unique set of `workclass` values.

```python
import pandas as pd

data = {'workclass': ['Private', 'Self-employed', 'Government', 'Private']}
df = pd.DataFrame(data)


unique_workclass = set(df['workclass'])


exists = "Self-employed" in unique_workclass
print(exists)  

workclass_list = ['Private', 'Self-employed', 'Government', 'Private']
unique_workclass = set(workclass_list)


exists = "Self-employed" in unique_workclass
print(exists)  
```

### **Exercise 6: Conditional Statements in Python**
**Question:** Write a function that returns "Minor" if the age is less than 18, "Adult" if between 18 and 65, and "Senior" if greater than 65. Test it on the first row's `age`.

```python
def age_category(age):
    if age < 18:
        return "Minor"
    elif 18 <= age <= 65:
        return "Adult"
    else:
        return "Senior"


first_row_age = 25  
print(age_category(first_row_age))  

```

### **Exercise 8: Functions**
**Question:** Define a function `sum_columns` that takes two column names and an index, sums the values of the specified columns at the given index, and returns the result. Test it with `age` and `hours-per-week` at the first row.

```python
import pandas as pd

data = {'age': [25, 30, 40], 'hours-per-week': [40, 50, 60]}
df = pd.DataFrame(data)

def sum_columns(df, col1, col2, index):
    return df.loc[index, col1] + df.loc[index, col2]

result = sum_columns(df, 'age', 'hours-per-week', 0)
print(result) 
```

### **Exercise 9: File Handling**

### **Exercise 10: Debugging in Python**
**Question:** Write a try/except block to handle a `KeyError` when accessing a non-existent column `"salary"`. Print an appropriate error message when caught.

```python
import pandas as pd

data = {'age': [25, 30, 40], 'workclass': ['Private', 'Self-employed', 'Government']}
df = pd.DataFrame(data)

try:
    # Attempt to access a non-existent column "salary"
    print(df['salary'])
except KeyError:
    print("Error: The column 'salary' does not exist in the DataFrame.")

```
